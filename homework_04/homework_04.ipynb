```
import datetime
from DataDescriptor import BaseDataDescriptor, OSDataDescriptor
from DataStream import BaseDataStream

class BaseFile:
    # можно обернуть функциями get/set при необходимости
    name:str
    size:int
    time_create:datetime
    owner:str

    _data_description:BaseDataDescriptor

    def save(self, stream : BaseDataStream)->bool:
        return stream.update_data(self, self._data_description)
    def load(self, stream : BaseDataStream)->bool:
        return stream.get_data(self, self._data_description)

class AudioFile(BaseFile):
    # можно вводить абстракцию кодека с enum
    format:str
    bitrate:int
    duration_sec:int

class VideoFile(BaseFile):
    # можно вводить абстракцию кодека с enum
    format:str
    resolution:list
    duration_sec:int

class PhotoFile(BaseFile):
    # можно вводить абстракцию кодека с enum
    format:str
    resolution:list
```
