# Задание 5 по линейной алгебре

```python
import numpy as np
import matplotlib as mpl
```

```python
array_size = 100

array_data = np.random.rand(array_size, array_size)
U, S, Vt = np.linalg.svd(array_data)
S = np.diag(S)

rank_error_lst = []
rank_range = range(2, array_size + 1)
for rank in rank_range:
    array_restore = U[:, :rank] @ S[:rank, :rank] @ Vt[:rank, :]
    rank_error_lst.append(np.linalg.norm(array_data - array_restore, ord="fro"))

mpl.pyplot.plot(rank_range, rank_error_lst, label="error")
mpl.pyplot.show()
```
