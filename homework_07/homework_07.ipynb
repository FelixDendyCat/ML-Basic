# I. Numpy

### Импортируйте NumPy

```python
import numpy as np
```

### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
array_linear = np.array([(1 if elem == 4 else 0) for elem in range(10)])
array_volume = array_linear.reshape(2, 5)
print(array_volume)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
array_direct = np.random.randint(10, 50, 10)
print(array_direct)

array_reverse = np.flip(array_direct)
print(array_reverse)

array_filtered = array_direct[array_direct % 2 == 0]
print(array_filtered)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
array_3_3 = np.random.randint(0, 9, (3,3))
print(array_3_3)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
array_dim = np.random.rand(4, 3, 2)
print(array_dim)
print(f"min: {array_dim.min()}, max: {array_dim.max()}")
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
array_6_4 = np.random.rand(6, 4)
#array_6_4 = np.array([1,2,3,4]).reshape(4, 1)
array_4_3 = np.random.rand(4, 3)
#array_4_3 = np.array([1,2,3,4]).reshape(1, 4)
array_mult = np.matmul(array_6_4, array_4_3)
print(array_mult)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
array_7_7 = np.random.rand(7, 7)
#array_7_7 = np.array([3,4,3,4]).reshape(2,2)
print(array_7_7)
print(f"mean: {np.mean(array_7_7)}, std: {np.std(array_7_7)}")
array_norm = (array_7_7 - np.mean(array_7_7)) / np.std(array_7_7)
print(array_norm)
```

# II. Pandas

### Импортируйте: pandas, matplotlib, seaborn

```python
import pandas as pd
import matplotlib as mpl
import seaborn as sns
```

### Загрузите датасет Tips из набора датасетов seaborn

```python
dataset_tips = sns.load_dataset("tips")
```

### Посмотрите на первые 5 строчек

```python
print(dataset_tips.head(5))
```

### Узнайте сколько всего строчек и колонок в данных

```python
print(dataset_tips.shape)
```

### Проверьте есть ли пропуски в данных

```python
dataset_tips.isnull().sum()
```

### Посмотрите на распределение числовых признаков

```python
print(dataset_tips.describe())
```

### Найдите максимальное значение 'total_bill'

```python
print(dataset_tips["total_bill"].max())
```

### Найдите количество курящих людей

```python
print(dataset_tips[dataset_tips["smoker"] == "Yes"].shape[0])
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
print(dataset_tips.groupby(["day"])["total_bill"].mean())
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_val = dataset_tips["total_bill"].median()
print(dataset_tips[dataset_tips["total_bill"] > median_val].groupby(["sex"])["tip"].mean())
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
replace_dict = {"No" : 0, "Yes" : 1}
dataset_tips["smoker"] = dataset_tips["smoker"].map(replace_dict)
dataset_tips.head(1)
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
sns.histplot(data=dataset_tips, x="total_bill")
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
sns.scatterplot(data=dataset_tips, x="total_bill", y="tip")
```

### Постройте pairplot

```python
sns.pairplot(dataset_tips)
mpl.pyplot.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
sns.boxplot(data=dataset_tips, y="total_bill", x="day")
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
fig, axes = mpl.pyplot.subplots(1, 2)

axes[0].set_title("Lunch")
sns.histplot(data = dataset_tips[dataset_tips["time"] == "Lunch"], x="tip", ax=axes[0])

axes[1].set_title("Dinner")
sns.histplot(data = dataset_tips[dataset_tips["time"] == "Dinner"], x="tip", ax=axes[1])

mpl.pyplot.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
fig, axes = mpl.pyplot.subplots(1, 2)

axes[0].set_title("Male")
sns.scatterplot(data = dataset_tips[dataset_tips["sex"] == "Male"],
                x = "total_bill", y = "tip", hue="smoker", ax=axes[0])

axes[1].set_title("Female")
sns.scatterplot(data = dataset_tips[dataset_tips["sex"] == "Female"],
                x = "total_bill", y = "tip", hue="smoker", ax=axes[1])

mpl.pyplot.show()
```
