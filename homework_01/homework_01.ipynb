## Задание 1: Зеркальное отражение цифр

Пользователь вводит пятизначное число. Программа должна зеркально отразить три центральные цифры. Первая и последняя остаются на местах.

<br>

**Примеры:**
* `23456 -> 25436`
* `30789 -> 38709`

```python
user_value_str = input()
# из какой позиции берем символы
char_pos = [0, 3, 2, 1, 4]
inverse_value_lst = [user_value_str[_pos] for _pos in char_pos]
inverse_value_str = ''.join(inverse_value_lst)
print(inverse_value_str)
```

## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
user_value_int = int(input())
# 2 дня за каждую полную неделю + 1 день за 6 дней неполной недели
holiday_count = (int(user_value_int / 7) * 2) + int((user_value_int % 7) / 6)
print(holiday_count)
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
# индексы в массиве исходных данных
INPUT_WIDTH = 0
INPUT_HEIGHT = 1
INPUT_SIZE = 2

# индексы в массиве размерностей
DIM_WIDTH = 0
DIM_HEIGHT = 1

# проверка разлома по всем измерениям
def CheckPossible(dim_lst:list, user_size:int, curr_size:int)->bool:
  posible_bool = False
  if (dim_lst[DIM_WIDTH] > 0 and dim_lst[DIM_HEIGHT] > 0) :
    for _dim in [DIM_WIDTH, DIM_HEIGHT] :
      posible_bool = CheckPossibleBreak(dim_lst[:], user_size, curr_size, _dim)
      if posible_bool:
        break
  return posible_bool

# проверка разлома по конкретному измерению
def CheckPossibleBreak(dim_lst:list, user_size:int, curr_size:int, break_dim:int)->bool:
  posible_bool = False
  if (break_dim == DIM_WIDTH):
    curr_size += dim_lst[DIM_WIDTH]
    dim_lst[DIM_HEIGHT] = dim_lst[DIM_HEIGHT] - 1
  elif (break_dim == DIM_HEIGHT) :
    curr_size += dim_lst[DIM_HEIGHT]
    dim_lst[DIM_WIDTH] = dim_lst[DIM_WIDTH] - 1
  if (curr_size == user_size) :
    posible_bool = True
  elif (curr_size > user_size) :
    posible_bool = False
  else :
    posible_bool = CheckPossible(dim_lst, user_size, curr_size)
  return posible_bool

user_data_str = input()
user_data_lst = [int(data) for data in user_data_str.split(',')]
possible = CheckPossible(user_data_lst[INPUT_WIDTH:INPUT_HEIGHT+1], user_data_lst[INPUT_SIZE], 0)
print(possible)
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
SYMB_PRINT = 0
SYMB_COST = 1
# Карта эквивалентов
symb_coll = [
  ('I', 1),
  ('V', 5),
  ('X', 10),
  ('L', 50),
  ('C', 100),
  ('D', 500),
  ('M', 1000),
]

dec_value_int = int(input())
roman_value_str = ""
for _symb_ind_inc in range(len(symb_coll) - 1, -1, -1):
  symb_cost_int = symb_coll[_symb_ind_inc][SYMB_COST]
  while dec_value_int >= symb_cost_int:
    roman_value_str += symb_coll[_symb_ind_inc][SYMB_PRINT]
    dec_value_int -= symb_cost_int
  # чем меньше вычтем, тем короче запись
  for _symb_ind_dec in range(_symb_ind_inc):
    dec_cost_int = symb_coll[_symb_ind_dec][SYMB_COST]
    sintetic_cost = symb_cost_int - dec_cost_int
    if (sintetic_cost != dec_cost_int and dec_value_int >= sintetic_cost):
      roman_value_str += symb_coll[_symb_ind_dec][SYMB_PRINT] + symb_coll[_symb_ind_inc][SYMB_PRINT]
      dec_value_int -= sintetic_cost
print(roman_value_str)
```

## Задание 5: Проверка на вещественное число

**Примеры:**
* `5.6  -> True`

```python
CHAR_CODE_0 = ord('0')
CHAR_CODE_9 = ord('9')
CHAR_CODE_POSITIVE = ord('+')
CHAR_CODE_NEGATIVE = ord('-')
CHAR_CODE_DECIMAL = ord('.')

correct_decimal_flag = True
negative_value_flag = False
decimal_value_flag = False

test_value_str = input()
for _char_ind in range(len(test_value_str)):
  char_code = ord(test_value_str[_char_ind])
  if char_code >= CHAR_CODE_0 and char_code < CHAR_CODE_9:
    continue
  elif char_code == CHAR_CODE_DECIMAL:
    if decimal_value_flag:
      correct_decimal_flag = False
      break
    decimal_value_flag = True
  elif char_code == CHAR_CODE_POSITIVE or char_code == CHAR_CODE_NEGATIVE:
    if _char_ind != 0 :
      correct_decimal_flag = False
      break
  else:
    correct_decimal_flag = False
    break
print(correct_decimal_flag)
```
