## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
def parse_snake(original_value:str)->list:
    delimiter_coll_lst = [_ind for _ind in range(len(original_value)) if original_value[_ind] == '_']
    # добавляем индексы -1 и len() для единообразия обработки
    delimiter_coll_lst.insert(0, -1)
    delimiter_coll_lst.append(len(original_value))
    word_coll_lst = [(delimiter_coll_lst[_ind - 1] + 1, delimiter_coll_lst[_ind]) for _ind in range(1, len(delimiter_coll_lst))]
    return word_coll_lst

def parse_camel(original_value:str)->list:
    delimiter_coll_lst = [_ind for _ind in range(len(original_value)) if original_value[_ind].isupper()]
    # добавляем индекс len() для единообразия обработки
    delimiter_coll_lst.append(len(original_value))
    word_coll_lst = [(delimiter_coll_lst[_ind - 1], delimiter_coll_lst[_ind]) for _ind in range(1, len(delimiter_coll_lst))]
    return word_coll_lst

def format_snake(original_value:str, word_coll_lst:list)->str:
    # по индексам выбираем строки, переводим в нижний регистр и объединяем через '_'
    return '_'.join([original_value[_word[0]:_word[1]].lower() for _word in word_coll_lst])

def format_camel(original_value:str, word_coll_lst:list)->str:
    # по индексам выбираем строки, делаем заглавной и объединяем
    return ''.join([original_value[_word[0]:_word[1]].title() for _word in word_coll_lst])

def modify_value(original_value:str, case_sign = None)->str:
    is_snake = original_value[0].islower() if len(original_value) > 0 else True
    word_coll_lst = parse_snake(original_value) if is_snake else parse_camel(original_value)

    to_snake = not is_snake
    if case_sign == 's':
        to_snake = True
    elif case_sign == 'C':
        to_snake = False
    return format_snake(original_value, word_coll_lst) if to_snake else format_camel(original_value, word_coll_lst)

#user_value_str = ['otus_course', 'PythonIsTheBest'][0]
user_value_str = input()
modified_value_str = modify_value(user_value_str)
print(modified_value_str)
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
DATA_COMPONENT_DAY = 0
DATA_COMPONENT_MONTH = 1
DATA_COMPONENT_YEAR = 2
DATA_COMPONENT_COUNT_ = 3

base_component_limit = [31, 12, 100500]
base_month_capacity = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31];

def check_user_input(user_value_str:str)->bool:
    valid_data = False
    component_str_value_lst = user_value_str.split('.')
    component_int_value_lst = [int(_component) for _component in component_str_value_lst if _component.isdecimal()]
    # проверяем общий формат: 3 числа разделенных точками
    if (len(component_str_value_lst) != DATA_COMPONENT_COUNT_):
        valid_data = False
    elif (len(component_str_value_lst) != DATA_COMPONENT_COUNT_):
        valid_data = False
    else:
        # проверяем значения на допуски: 1..31 день, 1..12 месяц, 1.._ год
        for _component_ind in range(DATA_COMPONENT_COUNT_):
            component_value_int = component_int_value_lst[_component_ind]
            if component_value_int <= 0:
                valid_data = False
                break
            elif component_value_int > base_component_limit[_component_ind]:
                valid_data = False
                break
        else:
            day_value = component_int_value_lst[DATA_COMPONENT_DAY]
            month_value = component_int_value_lst[DATA_COMPONENT_MONTH]
            year_value = component_int_value_lst[DATA_COMPONENT_YEAR]
            # считается ли год високосным
            long_year_bool = (year_value % 400) == 0 or ((year_value % 4) == 0 and (year_value % 100) != 0)
            # число дней в месяце
            month_capacity = base_month_capacity[month_value - 1]
            # число дней в феврале не високосного года
            if month_value == 2 and long_year_bool == False:
                month_capacity = 28
            # проверяем день месяца
            valid_data = False if day_value > month_capacity else True
    return valid_data

#user_value_str = ['29.02.2000', '29.02.2001', '31.04.1962'][2]
user_value_str = input()
valid_data = check_user_input(user_value_str)
print(valid_data)
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
import math

def check_prime_number(num:int)->bool:
    prime_num_bool = False
    if num < 0:
        prime_num_bool = False
    elif num < 2:
        prime_num_bool = True
    else:
        # делитель числа не может быть больше квадратного корня числа
        for _delimiter in range(2, int(math.sqrt(num)) + 1, 1):
            if (num % _delimiter) == 0:
                prime_num_bool = False
                break
        else:
            prime_num_bool = True
    return prime_num_bool

#user_value_int = [17, 20, 23][2]
user_value_int = int(input())
prime_num_bool = check_prime_number(user_value_int)
print(prime_num_bool)
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
INPUT_FNAME = 0
INPUT_SNAME = 1
INPUT_AGE = 2
INPUT_ID = 3
INPUT_COUNT = 4

USER_INPUT_OK = 0
USER_INPUT_ERROR_MASK_COUNT = 1 << 1
USER_INPUT_ERROR_MASK_FNAME = 1 << 2
USER_INPUT_ERROR_MASK_SNAME = 1 << 3
USER_INPUT_ERROR_MASK_AGE = 1 << 4
USER_INPUT_ERROR_MASK_ID = 1 << 5
USER_INPUT_ERROR_MASK_ID_UNIQUE = 1 << 6
def make_input_error_desc(error_mask:int)->str:
    input_error_desc = ''
    if (error_mask & USER_INPUT_ERROR_MASK_COUNT):
        input_error_desc += 'Ожидается ввод 4 параметров Имя Фамилия возраст ID\n'
    if (error_mask & USER_INPUT_ERROR_MASK_FNAME):
        input_error_desc += 'Имя должно состоять только из символов\n'
    if (error_mask & USER_INPUT_ERROR_MASK_SNAME):
        input_error_desc += 'Фамилия должна состоять только из символов\n'
    if (error_mask & USER_INPUT_ERROR_MASK_AGE):
        input_error_desc += 'Возраст должен быть числом от 18 до 60\n'
    if (error_mask & USER_INPUT_ERROR_MASK_ID):
        input_error_desc += 'ID - целое число, не более 8 знаков\n'
    if (error_mask & USER_INPUT_ERROR_MASK_ID_UNIQUE):
        input_error_desc += 'Пользователь с таким ID уже есть\n'
    return input_error_desc

# передаем list поскольку нам нужно отредактировать значение
def correct_name(user_input_lst:list, param_ind:int)->bool:
  is_correct = False
  if len(user_input_lst[param_ind]) <= 0:
    is_correct = False
  elif not user_input_lst[param_ind].isalpha():
    is_correct = False
  else:
    # проверка займет больше времени чем безусловная замена
    user_input_lst[param_ind] = user_input_lst[param_ind].title()
    is_correct = True
  return is_correct

# передаем list поскольку нам нужно отредактировать значение
def correct_age(user_input_lst:list, param_ind:int)->bool:
    is_correct = False
    if not user_input_lst[param_ind].isnumeric():
        is_correct = False
    else:
        age_int = int(user_input_lst[param_ind])
        if age_int < 18 or age_int > 60:
            is_correct = False
        else:
            user_input_lst[param_ind] = age_int
            is_correct = True
    return is_correct

# передаем list поскольку нам нужно отредактировать значение
def correct_id(user_input_lst:list, param_ind:int)->bool:
  is_correct = False
  if not user_input_lst[param_ind].isnumeric():
    is_correct = False
  else:
    id_int = int(user_input_lst[param_ind])
    if id_int > 100000000:
      is_correct = False
    else:
      user_input_lst[param_ind] = id_int
      is_correct = True
  return is_correct

def correct_user_input(base:dict, user_input_lst:list)->int:
    # оцениваем всю доступную информацию не прерываясь на первой ошибке
    user_input_error_mask = USER_INPUT_OK;
    if len(user_input_lst) != INPUT_COUNT:
        user_input_error_mask |= USER_INPUT_ERROR_MASK_COUNT
    if len(user_input_lst) > INPUT_FNAME:
        if not correct_name(user_input_lst, INPUT_FNAME):
            user_input_error_mask |= USER_INPUT_ERROR_MASK_FNAME
    if len(user_input_lst) > INPUT_SNAME:
        if not correct_name(user_input_lst, INPUT_SNAME):
            user_input_error_mask |= USER_INPUT_ERROR_MASK_SNAME
    if len(user_input_lst) > INPUT_AGE:
        if not correct_age(user_input_lst, INPUT_AGE):
            user_input_error_mask |= USER_INPUT_ERROR_MASK_AGE
    if len(user_input_lst) > INPUT_ID:
        if not correct_id(user_input_lst, INPUT_ID):
            user_input_error_mask |= USER_INPUT_ERROR_MASK_ID
        elif base.get(user_input_lst[INPUT_ID]) != None:
            user_input_error_mask |= USER_INPUT_ERROR_MASK_ID_UNIQUE
    return user_input_error_mask

USER_INFO_FNAME = 0
USER_INFO_SNAME = 1
USER_INFO_AGE = 2

def formatted_print_base(base:dict):
    # поиск максимальных строк для расчета ширины столбца
    # поскольку идет обращение к полю необходимо дефолтное значение
    max_fname_str = max(iter(base.values()), default = ('', '', 0), key = lambda _value: len(_value[USER_INFO_FNAME]))[USER_INFO_FNAME]
    max_sname_str = max(iter(base.values()), default = ('', '', 0), key = lambda _value: len(_value[USER_INFO_SNAME]))[USER_INFO_SNAME]
    # что бы не заводить новых констант, используем порядок ввода
    col_width_lst = [
        max(len(max_fname_str) + 1, 20),
        max(len(max_sname_str) + 1, 20),
        5,
        10
    ]

    # _item[0] - id, _item[1] - info
    sorted_base = sorted(base.items(), key = lambda _item: _item[1][USER_INFO_SNAME])

    # заголовок
    print(f"{'ID':{col_width_lst[INPUT_ID]}}", end = '')
    print(f"{'Фамилия':{col_width_lst[INPUT_SNAME]}}", end = '')
    print(f"{'Имя':{col_width_lst[INPUT_FNAME]}}", end = '')
    print(f"{'Возраст':{col_width_lst[INPUT_AGE]}}")

    # данные
    for _id, _info in sorted_base:
        # print(f"{f'${number:.3f}':>10s}")
        print(f"{f'{_id:08}':{col_width_lst[INPUT_ID]}}", end = '')
        print(f"{_info[USER_INFO_SNAME]:{col_width_lst[INPUT_SNAME]}}", end = '')
        print(f"{_info[USER_INFO_FNAME]:{col_width_lst[INPUT_FNAME]}}", end = '')
        print(f"{_info[USER_INFO_AGE]:{col_width_lst[INPUT_AGE]}}")

input_lst = [
#    'Иван1 Иванов1 3 2222222222222222 k',
#    'Иван Иванов 32 32',
    'Петр Петров 32 32',
    'Сидор Сидоров 33 33',
    'Петр Петров 35 35',
    'Иван Иванов 36 36',
    'Сидор Сидоров 37 37',
    'Петр Петров 38 38',
    '',
]
input_iterator = iter(input_lst)

base = dict()
input_not_empty_bool = True
while input_not_empty_bool:
    user_input_str = input()
    #user_input_str = next(input_iterator)
    input_not_empty_bool = user_input_str != ''
    if input_not_empty_bool:
        user_input_lst = user_input_str.split()
        user_input_error_mask = correct_user_input(base, user_input_lst)
        if user_input_error_mask == USER_INPUT_OK:
            base[user_input_lst[INPUT_ID]] = (user_input_lst[INPUT_FNAME], user_input_lst[INPUT_SNAME], user_input_lst[INPUT_AGE])
        else:
            print(make_input_error_desc(user_input_error_mask))

formatted_print_base(base)
```
